```python
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}

print(my_set)
```

```python
# Explanation:
# A set stores only unique values, so duplicate 'a' and 'i' are removed
```

```python
my_set[4] = 's'
```

```python
# Explanation:
# A set is unordered, so it doesn't have fixed positions/indexes like a list.
```

```python
set1 = {1, 3, 5, 7, 9}

set2 = {2, 3, 5, 8, 10}

print(set1)
print(set2)
```

```python
print("Union:", set1.union(set2))
```

```python
print("Intersection:", set1.intersection(set2))
```

```python
score = float(input("Enter your score (0 to 10): "))

if score < 0 or score > 10:
    print("Invalid score. Please enter a score between 0 and 10.")

elif score > 7:
    print("Above Average: Excellent work! Keep it up!")

elif score >= 4:
    print("Average: Good effort! Keep practicing, there's room for improvement.")

else:
    print("Below Average: Need to improve your performance. Consistent practice will lead to better results.")
```
