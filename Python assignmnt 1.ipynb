```python
string = "Hello Akshitha, welcome to Python programming"

# First character
print("First character:", string[0])

# Last character
print("Last character:", string[-1])

# First 5 characters
print("First 5 characters:", string[:5])

# Last 11 characters
print("Last 11 characters:", string[-11:])

# Reverse the string
print("Reverse:", string[::-1])

# Print the word Python
print("Word Python:", string[27:33])
```

```python
strM = "Python beginner tutorial"

# Convert to uppercase
print("Uppercase:", strM.upper())

# Convert to lowercase
print("Lowercase:", strM.lower())

# Capitalize the sentence
print("Capitalized:", strM.capitalize())

# Count the character 't'
print("Number of 't':", strM.count('t'))

# Replace Python with Machine Learning
print("Replaced:", strM.replace("Python", "Machine Learning"))
```

```python
# Creating two tuples
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

# Concatenate the two tuples
t_combine = tuple1 + tuple2
print("Combined tuple:", t_combine)

# Repeat t_combine 3 times
t_repeat = t_combine * 3
print("Repeated tuple:", t_repeat)

# Access the 3rd element
print("3rd element:", t_combine[2])

# Access the first three elements
print("First three elements:", t_combine[:3])

# Access the last three elements
print("Last three elements:", t_combine[-3:])
```
